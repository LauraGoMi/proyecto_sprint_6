```python
#Preparacion y limpieza de Datos
 #1. renombrar columnas a a minusculas
#==========================================
#CONVERTIR A MINUSCULAS LAS COLUMNAS
#==========================================

nombres_originales = df.columns.tolist()

df.columns = df.columns.str.lower().str.replace(' ', '_')

print("\n✅ Columnas renombradas exitosamente")
print("\nNombres ACTUALES:")
print(df.columns.tolist())

# Revisamos tipo de datos y convertimos

# user_score deberia tener datos numericos, no string, se hace el ajuste

# ==========================================
# CONVERTIR TIPOS DE DATOS
# ==========================================
# Verificar si hay valores 'tbd' (to be determined)
tbd_count = (df['user_score'] == 'tbd').sum()

print("\n Convirtiendo tipos de datos")
print("-" * 50)


#  Convertir user_score a flotat64

print("\n Convirtiendo 'user_score' a numérico...")

# Primero, reemplazamos 'tbd = to be determinated' con NaN (valor nulo)

df['user_score'] = df['user_score'].replace('tbd', np.nan)
print(f"   - Valores 'tbd' reemplazados por NaN")

# Ahora convertimos a numérico
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')
print(f"   ✅ Convertido. Tipo final: {df['user_score'].dtype}")
print(f"   Total valores nulos: {df['user_score'].isnull().sum()}")

print("\nTipos de datos FINALES:")
print(df.dtypes)

# ==========================================
# ANÁLISIS DE VALORES AUSENTES
# ==========================================

print("\nAnalizando valores ausentes")
print("-" * 50)

# Crear un resumen completo de valores nulos
valores_nulos = pd.DataFrame({
    'Columna': df.columns,
    'Valores_Nulos': df.isnull().sum(),
    'Porcentaje': (df.isnull().sum() / len(df) * 100).round(2)
})

# Ordenar por porcentaje descendente para ver las columnas más problemáticas
valores_nulos = valores_nulos.sort_values('Porcentaje', ascending=False)

print("\nResumen de valores ausentes:")
print(valores_nulos)


plt.figure(figsize=(12, 6))
valores_nulos_plot = valores_nulos[valores_nulos['Valores_Nulos'] > 0]

# Crear gráfico de barras
plt.barh(valores_nulos_plot['Columna'], valores_nulos_plot['Porcentaje'], color='coral')
plt.xlabel('Porcentaje de valores nulos (%)', fontsize=12)
plt.ylabel('Columnas', fontsize=12)
plt.title('Porcentaje de Valores Nulos por Columna', fontsize=14, fontweight='bold')
plt.xlim(0, 100)

# Agregar valores en las barras
for index, value in enumerate(valores_nulos_plot['Porcentaje']):
    plt.text(value + 1, index, f'{value}%', va='center')

plt.tight_layout()
plt.show()



# ==========================================
# ESTRATEGIA
# ==========================================

print("\n VALORES AUSENTES")
print("-" * 50)

# Vamos a analizar cada columna con valores nulos y decidir qué hacer

print("\n ANÁLISIS POR COLUMNA (para las mas representativas en valores nulos):\n")

# --- YEAR_OF_RELEASE ---
print("1. YEAR_OF_RELEASE:")
print(f"   - Valores nulos: {df['year_of_release'].isnull().sum()}")

# --- CRITIC_SCORE ---
print("2. CRITIC_SCORE:")
print(f"   - Valores nulos: {df['critic_score'].isnull().sum()} ({(df['critic_score'].isnull().sum()/len(df)*100):.1f}%)")

# --- USER_SCORE ---
print("3.USER_SCORE:")
print(f"   - Valores nulos: {df['user_score'].isnull().sum()} ({(df['user_score'].isnull().sum()/len(df)*100):.1f}%)")


# --- RATING (ESRB) ---
print("4. RATING:")
print(f"   - Valores nulos: {df['rating'].isnull().sum()} ({(df['rating'].isnull().sum()/len(df)*100):.1f}%)")

# Opcional: Crear categoría 'Unknown' para rating
# df['rating'] = df['rating'].fillna('Unknown')


```

CONCLUSIONES:

- POR COLUMNAS (representativas):

* YEAR_OF_RELEASE
- se van a mantener los valores nulos con NaN, ya que el año es indispensable para saber la informacion real y confiable, mas adelante lo que se hace es que no se toman en cuenta para no distorsionar la informacion.

*CRITIC_SCORE

- Mantener como NaN, ya que todos los juegos son evaluados por criticos del segmento y si no son evaluados es porque son poco conocidos o muy antiguos y por esto seria informacion irrelevante de analizar, si se rellena con un promedio la informacion podria dejar de ser confiable.

*USER_SCORE :
- Mantener como NaN, es un caso similar a critic_score, ya convertidos los to be confirmated (tbd) a NaN, puede ser informacion muy antigua o juegos poco conocidos que no ameritan una calificacion de usuario.


*RATING (ESRB) :
- Mantener como NaN, ya que se considera que no todas las regiones requieren calificacion ESRB, por juegos sin region o muy antiguos.

CONCLUSIONES GENERALES:

1. NO rellenamos valores numéricos (critic_score, user_score) porque:
   - Distorsionaria la informacion con datos irreales.
   - Son valores que no existen,  no por errores de captura sino por otras razones, por ejemplo, informacion muy antigua. 
   
2. NO eliminamos filas porque:
   - Perdríamos información valiosa de ventas
   - Los valores nulos son informativos por sí mismos
   
3. En análisis específicos, filtraremos según necesidad:
   - Para correlaciones: solo registros CON scores
   - Para análisis temporal: solo registros CON año

```python
# ==========================================
# DIAGRAMA DE CAJA PARA EL TOP 5 PLATAFORMAS SELECCIONADAS EN EL PASO ANTERIOR
# ==========================================

print("ANÁLISIS DETALLADO TOP 5 PLATAFORMAS")
print("="*60)

# Guardar nombres de las plataformas
top_5_nombres = list(top_5_plataformas)  

# Filtrar dataset solo con TOP 5
df_top5 = df_relevante[df_relevante['platform'].isin(top_5_nombres)].copy()

print(f"\nDataset TOP 5: {len(df_top5):,} juegos")
print(f"   Representa el {(len(df_top5)/len(df_relevante)*100):.1f}% de los juegos del periodo")

# -------------------------------------
# ESTADÍSTICAS DESCRIPTIVAS TOP 5
# -------------------------------------
estadisticas = []

for plataforma in top_5_nombres:
    ventas_plat = df_top5[df_top5['platform'] == plataforma]['total_sales']
    
    stats = {
        'Plataforma': plataforma,
        'Juegos': len(ventas_plat),
        'Ventas_Totales': ventas_plat.sum(),
        'Promedio': ventas_plat.mean(),
        'Mediana': ventas_plat.median(),
        'Desv_Std': ventas_plat.std(),
        'Min': ventas_plat.min(),
        'Max': ventas_plat.max()
    }
    estadisticas.append(stats)

df_stats = pd.DataFrame(estadisticas)

# -------------------------------------
# BOXPLOT PRINCIPAL
# -------------------------------------

plt.figure(figsize=(14, 8))

# Preparar datos para boxplot
# ¿Qué hace esto? Crea una lista con las ventas de cada plataforma
box_data = [df_top5[df_top5['platform'] == plat]['total_sales'].values 
            for plat in top_5_nombres]

# Crear boxplot
bp = plt.boxplot(box_data, 
                 labels=top_5_nombres,
                 patch_artist=True,      # Permite colorear las cajas
                 showfliers=True,        # Muestra outliers (valores extremos)
                 notch=False,            
                 widths=0.6,
                 boxprops=dict(linewidth=1.5),
                 medianprops=dict(color='red', linewidth=3),
                 whiskerprops=dict(linewidth=1.5),
                 capprops=dict(linewidth=1.5))

# Colorear las cajas con colores diferentes
colores = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
for patch, color in zip(bp['boxes'], colores):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

# Etiquetas y título
plt.xlabel('Plataforma', fontsize=13, fontweight='bold')
plt.ylabel('Ventas Globales por Juego (Millones $)', fontsize=13, fontweight='bold')
plt.title('Distribución de Ventas Globales - TOP 5 Plataformas (2009-2016)\nAnálisis de Rendimiento y Valores Atípicos', 
          fontsize=15, fontweight='bold', pad=20)
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.xticks(fontsize=12, fontweight='bold')

# Agregar nota explicativa
plt.text(0.02, 0.98, 'Línea roja = Mediana\nPuntos = Outliers (juegos excepcionales)', 
         transform=plt.gca().transAxes, fontsize=10, 
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()


# -------------------------------------
# ANÁLISIS: PROMEDIO VS MEDIANA
# -------------------------------------

print("\nANÁLISIS: PROMEDIO VS MEDIANA")
print("="*60)

comparacion = df_stats[['Plataforma', 'Promedio', 'Mediana']].copy()
comparacion['Diferencia'] = comparacion['Promedio'] - comparacion['Mediana']
comparacion['Dif_%'] = ((comparacion['Diferencia'] / comparacion['Mediana']) * 100).round(1)

comparacion = comparacion.sort_values('Promedio', ascending=False)

print("\nTabla comparativa:")
print(comparacion.to_string(index=False))
```

```python
# ==========================================
# CORRELACIÓN REVIEWS Y VENTAS
# ==========================================

# SELECCIONAR 1 PLATAFORMA POPULAR

# Voy a elegir la plataforma #1 en ventas para análisis detallado
plataforma_analisis = top_5_nombres[0]

print(f"\nPlataforma seleccionada: {plataforma_analisis}")
print(f"Razón: Plataforma líder en ventas del TOP 5")

# Filtrar datos de esta plataforma
df_plat = df_top5[df_top5['platform'] == plataforma_analisis].copy()

# Crear los datasets filtrados para análisis de correlación
df_plat_critic = df_plat[df_plat['critic_score'].notna()].copy()
df_plat_user = df_plat[df_plat['user_score'].notna()].copy()

print(f"Juegos con critic_score: {len(df_plat_critic)}")
print(f"Juegos con user_score: {len(df_plat_user)}")
# -------------------------------------
# ANÁLISIS 1: CRITIC_SCORE VS VENTAS
# -------------------------------------

print("\n" + "="*60)
print(f" ANÁLISIS 1: CRITIC_SCORE VS VENTAS ({plataforma_analisis})")
print("="*60)

# Calcular correlación
correlacion_critic = df_plat_critic['critic_score'].corr(df_plat_critic['total_sales'])

print(f"\nCOEFICIENTE DE CORRELACIÓN: {correlacion_critic:.4f}")

# -------------------------------------
# GRÁFICO:  CRITIC_SCORE
# -------------------------------------
plt.figure(figsize=(12, 7))

# Scatter plot

plt.scatter(df_plat_critic['critic_score'], 
            df_plat_critic['total_sales'],
            alpha=0.5,           # Transparencia para ver puntos superpuestos
            s=50,                # Tamaño de puntos
            color='steelblue',
            edgecolors='black',
            linewidth=0.5)

z = np.polyfit(df_plat_critic['critic_score'], df_plat_critic['total_sales'], 1)
p = np.poly1d(z)
plt.plot(df_plat_critic['critic_score'], 
         p(df_plat_critic['critic_score']), 
         "r--", 
         linewidth=2, 
         label=f'Línea de tendencia\nCorrelación: {correlacion_critic:.3f}')

plt.xlabel('Critic Score (0-100)', fontsize=13, fontweight='bold')
plt.ylabel('Ventas Globales (Millones $)', fontsize=13, fontweight='bold')
plt.title(f'Relación entre Critic Score y Ventas - {plataforma_analisis} (2009-2016)\nAnálisis de Correlación', 
          fontsize=15, fontweight='bold', pad=20)
plt.legend(fontsize=11, loc='upper left')
plt.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

# -------------------------------------
# ANÁLISIS 2: USER_SCORE VS VENTAS
# -------------------------------------

print("\n" + "="*60)
print(f"ANÁLISIS 2: USER_SCORE VS VENTAS ({plataforma_analisis})")
print("="*60)

# Calcular correlación
correlacion_user = df_plat_user['user_score'].corr(df_plat_user['total_sales'])

print(f"\n COEFICIENTE DE CORRELACIÓN: {correlacion_user:.4f}")


# -------------------------------------
# GRÁFICO:  USER_SCORE
# -------------------------------------

plt.figure(figsize=(12, 7))

# Scatter plot
plt.scatter(df_plat_user['user_score'], 
            df_plat_user['total_sales'],
            alpha=0.5,
            s=50,
            color='coral',
            edgecolors='black',
            linewidth=0.5)

# Línea de tendencia
z_user = np.polyfit(df_plat_user['user_score'], df_plat_user['total_sales'], 1)
p_user = np.poly1d(z_user)
plt.plot(df_plat_user['user_score'], 
         p_user(df_plat_user['user_score']), 
         "r--", 
         linewidth=2, 
         label=f'Línea de tendencia\nCorrelación: {correlacion_user:.3f}')

# Etiquetas
plt.xlabel('User Score (0-10)', fontsize=13, fontweight='bold')
plt.ylabel('Ventas Globales (Millones $)', fontsize=13, fontweight='bold')
plt.title(f'Relación entre User Score y Ventas - {plataforma_analisis} (2009-2016)\nAnálisis de Correlación', 
          fontsize=15, fontweight='bold', pad=20)
plt.legend(fontsize=11, loc='upper left')
plt.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

- Se puede identificar que la correlacion entre Critic Score vs Ventas es del 0.41, es decir una correlacion Moderada, con lo que podemos apreciar que los consumidores o usuarios "algunos" revisan los reviews de criticos y personas con experiencia para tomar la determinacion de compra.

- Por otro lado la correlacion entre User Score vs Ventas es de 0.09, es decir casi nula, lo que nos lleva a pensar que los usuarios no revisan o toman en cuenta los reviews de otros usuarios, esto seguramente es dado a que los videjuegos tienen que ver mucho con la personalidad de cada ser y sus gustos, es algo subjetivo.

```python
# ==========================================
# PASO 5:PRUEBAS DE HIPÓTESIS
# ==========================================

print("PASO 5: PRUEBAS DE HIPÓTESIS ESTADÍSTICAS")
print("="*60)

# Importar librería para pruebas estadísticas
from scipy import stats

# Definimos nivel de significancia
alpha = 0.05
print(f"✅ Nivel de significancia establecido: alpha = {alpha}")
print(f"   (Usamos 0.05 por que es estandar)\n")

# ==========================================
# HIPÓTESIS 1: XBOX ONE vs PC
# ==========================================

print("\n" + "="*60)
print("HIPÓTESIS 1: Calificaciones de Xbox One vs PC")
print("="*60)

# -------------------------------------
# FORMULACIÓN DE HIPÓTESIS
# -------------------------------------

print("\nFORMULACIÓN DE HIPÓTESIS")
print("-"*60)

print("""
H0 (Hipótesis Nula): 
   Las calificaciones promedio de usuarios para Xbox One y PC son IGUALES.

H1 (Hipótesis Alternativa):
   Las calificaciones promedio de usuarios para Xbox One y PC son DIFERENTES.
""")

# -------------------------------------
# PREPARAR DATOS
# -------------------------------------

print("\nPREPARACIÓN DE DATOS: user_score")
print("-"*60)

# Filtrar datos de Xbox One
xone_scores = df_relevante[df_relevante['platform'] == 'XOne']['user_score'].dropna()

# Filtrar datos de PC
pc_scores = df_relevante[df_relevante['platform'] == 'PC']['user_score'].dropna()

print(f"  Xbox One: {len(xone_scores)} calificaciones")
print(f"  PC:       {len(pc_scores)} calificaciones")

# Verificar que hay suficientes datos
if len(xone_scores) < 30 or len(pc_scores) < 30:
    print("\nOJO: Una o ambas muestras tienen menos de 30 observaciones")
    print("   Los resultados pueden no ser totalmente confiables")

# Estadísticas descriptivas
print(f"\nEstadísticas descriptivas:")
print(f"  Xbox One - Promedio: {xone_scores.mean():.3f} | Mediana: {xone_scores.median():.3f}")
print(f"  PC       - Promedio: {pc_scores.mean():.3f} | Mediana: {pc_scores.median():.3f}")

# -------------------------------------
# PRUEBA ESTADÍSTICA
# -------------------------------------

# Realizar la prueba t
statistic, p_value = stats.ttest_ind(xone_scores, pc_scores, equal_var=False)

print(f"\nRESULTADOS DE LA PRUEBA:")
print("-"*60)
print(f"P-value:       {p_value:.4f}")
print(f"Alpha (α):     {alpha}")

# -------------------------------------
# INTERPRETACIÓN
# -------------------------------------

print("\nINTERPRETACIÓN Y DECISIÓN")
print("-"*60)

if p_value < alpha:
    decision = "RECHAZAMOS H0"
    conclusion = "Las calificaciones promedio de Xbox One y PC son  DIFERENTES"
    
else:
    decision = "APROBAMOS H0"
    conclusion = "Las calificaciones promedio de Xbox One y PC son  IGUALES"
   

print(f"\n DECISIÓN: {decision}")
print(f"\nCONCLUSIÓN:")
print(f"   {conclusion}")


# ==========================================
# HIPÓTESIS 2: ACTION vs SPORTS
# ==========================================

print("\n" + "="*60)
print("**HIPÓTESIS 2**: Calificaciones de Action vs Sports")
print("="*60)

# -------------------------------------
# FORMULACIÓN DE HIPÓTESIS
# -------------------------------------

print("\n FORMULACIÓN DE HIPÓTESIS")
print("-"*60)

print("""
H0 (Hipótesis Nula):
   Las calificaciones promedio de usuarios para Action y Sports son IGUALES.

H1 (Hipótesis Alternativa):
   Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son DIFERENTES
""")

# -------------------------------------
# PREPARAR DATOS
# -------------------------------------

print("\nPREPARACIÓN DE DATOS")
print("-"*60)

# Filtrar datos de Action
action_scores = df_relevante[df_relevante['genre'] == 'Action']['user_score'].dropna()

# Filtrar datos de Sports
sports_scores = df_relevante[df_relevante['genre'] == 'Sports']['user_score'].dropna()

print(f"\nDatos disponibles:")
print(f"  Action: {len(action_scores)} calificaciones")
print(f"  Sports: {len(sports_scores)} calificaciones")

# Verificar suficiencia de datos
if len(action_scores) < 30 or len(sports_scores) < 30:
    print("\nOJO: Una o ambas muestras tienen menos de 30 observaciones")
else:
    print("\n✅ Ambas muestras tienen suficientes datos (n ≥ 30)")

# Estadísticas descriptivas
print(f"\n Estadísticas descriptivas:")
print(f"  Action - Promedio: {action_scores.mean():.3f} | Mediana: {action_scores.median():.3f}")
print(f"  Sports - Promedio: {sports_scores.mean():.3f} | Mediana: {sports_scores.median():.3f}")

# -------------------------------------
# PRUEBA ESTADÍSTICA
# -------------------------------------

# Realizar la prueba t
statistic2, p_value2 = stats.ttest_ind(action_scores, sports_scores, equal_var=False)

print(f"\nRESULTADOS DE LA PRUEBA:")
print("-"*60)
print(f"P-value:       {p_value2:.4f}")
print(f"Alpha (α):     {alpha}")

# -------------------------------------
# INTERPRETACIÓN
# -------------------------------------

print("\nINTERPRETACIÓN Y DECISIÓN")
print("-"*60)

if p_value2 < alpha:
    decision2 = "RECHAZAMOS H0"
    conclusion2 = "Las calificaciones promedio de Action y Sports son DIFERENTES"
    
else:
    decision2 = "APROBAMOS H0"
    conclusion2 = "Las calificaciones promedio de Action y Sports son IGUALES"
   
print(f"\n DECISIÓN: {decision2}")
print(f"\nCONCLUSIÓN:")
print(f"   {conclusion2}")
print(f"\nEXPLICACIÓN:")
if p_value2 < alpha:
    print(f"   • P-value ({p_value2:.4f}) < α ({alpha})")
    print(f"   • La probabilidad de que esta diferencia sea por azar es < {alpha*100}%")
    print(f"   • Conclusión: La diferencia es REAL y estadísticamente significativa")
else:
    print(f"   • P-value ({p_value2:.4f}) ≥ α ({alpha})")
    print(f"   • La probabilidad de que esta diferencia sea por azar es ≥ {alpha*100}%")
    print(f"   • Conclusión: NO podemos afirmar que hay diferencia real")


```
